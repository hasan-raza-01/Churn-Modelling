### Buisness Objective
The objective of this analysis is to understand the factors affecting churn of account holders. By understanding resons behind why customer moves towards churn allows to create retention plans, adjust pricing, improve service on the high risk segment regions using `shubh0799/churn-modelling` dataset from kaggle.

### Data Overview 
There are `10,000` records with demographic details( `Surname`, `Gender`, `Age`, `Geography`, `EstimatedSalary` ), account details( `CustomerId`, `CreditScore`, `Tenure`, `Balance`, `NumOfProducts`, `HasCrCard`, `IsActiveMember` ) and target( `Exited` ) with zero incomplete records. 

|Feature            |Type       |Description
|-------------------|-----------|------------
|`CustomerId`       |`integer`  |Unique Identification Number for customer provided by bank.
|`Surname`          |`string`   |Last name of Customer.
|`CreditScore`      |`integer`  |Credit Score of the account holder.
|`Geography`        |`string`   |Country where the account holder is having account. 
|`Gender`           |`string`   |Gender of the account holder. 
|`Age`              |`integer`  |Age of the account holder.
|`Tenure`           |`integer`  |Total number of years account holder is connected.
|`Balance`          |`float`    |current balance of account holder.
|`NumOfProducts`    |`integer`  |Number of bank product/services account holder is connected.
|`HasCrCard`        |`integer`  |`0` for customer not carries bank credit card, `1` for customer carries bank credit card.
|`IsActiveMember`   |`integer`  |`0` for customer is not an active member, `1` for customer is an active number.
|`EstimatedSalary`  |`float`    |Estimated salary of the account holder.
|`Exited`           |`integer`  |`0` for customer retained, `1` for customer churned.

### **Outliers and Anomalies**

After calculating higher fence through Quartiles and InterQuartileRange we get 62 and any datapoint above 62 is an outlier but it is not as higher age leads to more churn. So its not an outlier but a buisness anomaly which needs to handled carefully rather than removing.

### **Simpson's Paradox**

When segmenting Balance and Churn Rate by NumOfProducts a Paradox was found. As higher average balance and higher number of products holding leading to higher churn rate but in this segmentation it has been observed that though customers holding 1 products having highest average balance than customers holding 3 and 4 and having a lower churn rate than those category. However customers holding more products have higher churn rate but customers holding 2 products have lower churn rate than customers holding 1 product through it also have a lower avg balance but if bank will aggresively pushes more and more products towards customers holding lower number of product thinking it will increase retention then it is simply going to increase churn rate rathen then retention.

### **Geographic Analysis**

In terms of geographic region Germany is having 32% of churn rate where France and Spain are having churn rate of 16% and 17% respectively. Germany also having customers with higher age and higher balance which are highly co-related to churn.

### **Recommendations**

1. As Germany is have customers with higher age along with higher balance and higher churn rate than france or spain must needs an investigation on region to find out wheather its a geographic region or service issue.
2. Must focus on making process easier to retain customers with older demographics. 
3. We should introduce retention programes for females. 
4. Customers holding 3 and 4 product have higher churn rate which indicates towards poor service.
5. Definitly don't push aggresively more products to customers as churn is higher in customers holding more number of products.
